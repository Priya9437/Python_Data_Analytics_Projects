```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("--- PRIYA'S ZOMATO PROJECT IS LIVE ---")
```

Step - 2 Create the data frame

```python
# तुम्हारी अपनी रीब्रैंडेड फ़ाइल 'Zomato_Raw_Data.csv' के साथ डेटा लोड करना
df = pd.read_csv('Zomato_Raw_Data.csv', encoding='unicode_escape')

# चेक करना कि टेबल मख्खन की तरह लोड हुई या नहीं (शुरू की 5 लाइनें)
print(df)
```

```python
df
```

Convert the dat type of column - Rate

```python
def handleRate(value):
    if (value == 'NEW' or value == '-'):
        return np.nan
    else:
        value = str(value).split('/')
        value = value[0] # स्लैश के पहले का साफ़ नंबर
        return float(value)

# मशीन को रेट कॉलम पर चलाया
df['rate'] = df['rate'].apply(handleRate)

# 👑 अब पूरी की पूरी चमचमाती टेबल देखने के लिए सिर्फ यह लिखो:
df.head()
```

Type of Resturent

```python
df.head()
```

```python
sns.countplot(x = df['listed_in(type)'])
plt.xlabel("type of resturent")
```

Conclusion - majority of the resturent falls in dinning category

```python
df.head()
```

```python
grouped_data = df.groupby('listed_in(type)')['votes'].sum()

result = pd.DataFrame({'votes': grouped_data})
plt.plot(result, c="green", marker="o")
plt.xlabel("Type of restaurant", c="red", size=20)
plt.ylabel("Votes", c="red", size=20)
```

Conclusion - dinning resturents has received maximum votes

```python
df.head()
```

```python
plt.hist(df['rate'], bins=5)
plt.title("Distribution of Ratings")
plt.show()
```

Conclusion - majority of the resturent received ratings from 3.5 to 4.5

```python
df.head()
```

```python
couple_data = df['approx_cost(for two people)']
sns.countplot(x=couple_data)
```

Conclusion - majority of couples preferr resturents with an approximate cost of 300 rs. 

Which mode receives maximum ratings

```python
df.head()
```

```python
plt.figure(figsize = (6,6))
sns.boxplot(x = 'online_order', y = 'rate', data = df)
```

Conclusion - offline order received lower rating in comparison to online.

```python
pivot_table = df.pivot_table(index='listed_in(type)', columns='online_order', aggfunc='size', fill_value=0)
sns.heatmap(pivot_table, annot=True, cmap='YlGnBu', fmt='d')
plt.title("Heatmap")
plt.xlabel("Online Order")
plt.ylabel("Listed In (Type)")
plt.show()
```

Conclusion - Dinning resturents primarily accepts offline orders wherease cafes primarily receive online orders. This suggests clients
preferrs to order in person at resturent, but prefer online ordering in cafes.
